```python
from modeling_amt.act_utils import ACT_basic, gen_timing_signal
import torch
from torch import nn
```

```python
class Model(nn.Module):
    def __init__(self, hidden_size, num_layers, max_length):
        super().__init__()
        self.emb = nn.Embedding(10, hidden_size)
        self.block = nn.TransformerEncoderLayer(d_model=hidden_size, dim_feedforward=hidden_size * 4, nhead=4, batch_first=True)
        self.act = ACT_basic(hidden_size)
        self.time_enc = gen_timing_signal(max_length, hidden_size)
        self.pos_enc = gen_timing_signal(num_layers, hidden_size)
    def forward(self, x):
        x = self.emb(x)
        mask = nn.Transformer.generate_square_subsequent_mask(x.size(-2))
        x = self.act(x, x, lambda x: self.block(x, src_mask=mask, is_causal=True), self.time_enc, self.pos_enc, max_hop=4)
        return x

model = Model(16, 10, 100).to('cuda')
x = torch.randn((5, 6, 16))
t = torch.randint(0, 10, (5, 6)).to('cuda')
mask = nn.Transformer.generate_square_subsequent_mask(x.size(-2))
out, state = model(t)
```

```python
import torch
import torch.nn as nn
from functools import partial
from fvcore.nn import FlopCountAnalysis

# Example model with multiple inputs
class MultiInputModel(nn.Module):
    def __init__(self):
        super(MultiInputModel, self).__init__()
        self.embedding = nn.Embedding(30522, 128)
        self.attention = nn.Linear(128, 128)
        self.fc = nn.Linear(128, 2)

    def forward(self, input_ids, attention_mask=None):
        x = self.embedding(input_ids)
        if attention_mask is not None:
            x = x * attention_mask.unsqueeze(-1)  # Masking
        x = self.attention(x)
        x = x.mean(dim=1)  # Pooling
        x = self.fc(x)
        return x

# Initialize model and inputs
model = MultiInputModel()
input_ids = torch.randint(0, 30522, (1, 20))  # Example input_ids
attention_mask = torch.ones(1, 20)  # Example attention_mask

# Create a partial function to handle multiple inputs
partial_model = partial(model.forward, input_ids=input_ids, attention_mask=attention_mask)

# FLOPs analysis
flop_analysis = FlopCountAnalysis(model, (input_ids, attention_mask))
print(f"FLOPs: {flop_analysis.total()}")  # Total FLOPs
```

```python
from ptflops import get_model_complexity_info
import torch
import torch.nn as nn

# Original model with multiple inputs
class MultiInputModel(nn.Module):
    def __init__(self):
        super(MultiInputModel, self).__init__()
        self.embedding = nn.Embedding(30522, 128)
        self.attention = nn.Linear(128, 128)
        self.fc = nn.Linear(128, 2)

    def forward(self, input_ids, attention_mask=None):
        x = self.embedding(input_ids)
        if attention_mask is not None:
            x = x * attention_mask.unsqueeze(-1)  # Masking
        x = self.attention(x)
        x = x.mean(dim=1)  # Pooling
        x = self.fc(x)
        return x

# Wrapper model for compatibility with ptflops
class WrapperModel(nn.Module):
    def __init__(self, model):
        super(WrapperModel, self).__init__()
        self.model = model

    def forward(self, inputs):
        input_ids, attention_mask = inputs
        return self.model(input_ids, attention_mask)

# Initialize models
original_model = MultiInputModel()
wrapped_model = WrapperModel(original_model)

# Define input constructor for ptflops
def input_constructor(dummy_input):
    input_ids = torch.randint(0, 30522, (1, 20))  # Example input_ids
    attention_mask = torch.ones(1, 20)  # Example attention_mask
    return (input_ids, attention_mask)  # Return as a tuple

# Compute FLOPs and parameters
macs, params = get_model_complexity_info(
    wrapped_model,
    (1, 20),  # Placeholder for input shape
    as_strings=True,
    input_constructor=input_constructor,
    verbose=True,
)

print(f"MACs (Multiply-Accumulate Operations): {macs}")
print(f"Parameters: {params}")
```
