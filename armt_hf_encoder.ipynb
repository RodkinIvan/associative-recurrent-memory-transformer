```python
from transformers import AutoModelForCausalLM

repo_id = "XXXX/armt-neox-tiny"
loaded = AutoModelForCausalLM.from_pretrained(repo_id, trust_remote_code=True)
print(type(loaded))
```

```python
armt_cfg = type(loaded.config)(
    base_model_config=None, # if you want to create model from scratch
    base_model_name='prajjwal1/bert-tiny', # if you want to use pretrained model
    num_mem_tokens=16,
    d_mem=32,
    segment_size=128,
    segment_alignment="left",
    sliding_window=False,
    layers_attr="bert.encoder.layer",
    wrap_pos=False,
    correction=True,
    n_heads=1,
    use_denom=True,
    gating=False,
    freeze_mem=False,
    act_on=False,
)


bert_model = type(loaded)(armt_cfg)
```

```python
import torch
input_ids = torch.randint(0, 100, (5, 200))
bert_model(input_ids=input_ids, attention_mask=torch.ones_like(input_ids)).logits.shape
```
